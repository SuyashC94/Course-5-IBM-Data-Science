```python
import pandas as pd
import plotly.graph_objects as go
import yfinance as yf

tsla = yf.Ticker('TSLA')

tesla_data = tsla.history(period = 'max')
tesla_data
```

```python
tesla_data.head()
```

```python
tesla_data.tail()
```

```python
fig = go.Figure(data = go.Scatter(x=tesla_data.index, y=tesla_data['Close']))
fig.update_layout(title = 'TESLA STOCK FROM START TO TODAY')
fig.show()
```

```python
from plotly.subplots import make_subplots
fig2 = make_subplots(specs=[[{'secondary_y': True}]])
fig2.add_trace(go.Scatter(x=tesla_data.index, y=tesla_data['Close'], name='Revenue'),secondary_y=False)
fig2.update_layout(title='Tesla Revenue Chart')
fig2.show()
```
